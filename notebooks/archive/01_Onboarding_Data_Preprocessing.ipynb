## 13. Kenyan Prototype Adaptation

The source dataset represents synthetic digital onboarding applications in Nigeria.

## 14. Task Type Creation below

The original KYC tiers were converted into understandable onboarding task categories.

- Tier 1 was represented as Basic Customer Onboarding.
- Tier 2 was represented as Standard KYC Onboarding.
- Tier 3 was represented as Enhanced KYC Onboarding.

The tiers were also mapped to Low, Medium and High task complexity respectively.

These labels are project-specific interpretations and are not presented as official Kenyan regulatory KYC classifications.

```python
task_type_mapping = {
    "tier1": "Basic Customer Onboarding",
    "tier2": "Standard KYC Onboarding",
    "tier3": "Enhanced KYC Onboarding"
}

tasks["task_type"] = tasks["kyc_tier"].map(
    task_type_mapping
)
```

```python
complexity_mapping = {
    "tier1": "Low",
    "tier2": "Medium",
    "tier3": "High"
}

tasks["task_complexity"] = tasks["kyc_tier"].map(
    complexity_mapping
)
```

```python
tasks[
    ["kyc_tier", "task_type", "task_complexity"]
].head(10)
```

```python
id_threshold = tasks[
    "id_verification_score"
].quantile(0.25)

face_threshold = tasks[
    "face_match_score"
].quantile(0.25)

print("ID verification threshold:", id_threshold)
print("Face match threshold:", face_threshold)
```

```python
tasks["needs_id_review"] = (
    tasks["id_verification_score"] < id_threshold
)

tasks["needs_face_review"] = (
    tasks["face_match_score"] < face_threshold
)
```

```python
tasks[
    [
        "id_verification_score",
        "face_match_score",
        "needs_id_review",
        "needs_face_review"
    ]
].head(10)
```

```python
def generate_required_skills(row):

    skills = ["Customer Onboarding"]

    if row["kyc_tier"] in ["tier2", "tier3"]:
        skills.append("KYC Verification")

    if row["kyc_tier"] == "tier3":
        skills.append("Enhanced KYC Review")

    if row["needs_id_review"]:
        skills.append("Identity Verification")

    if row["needs_face_review"]:
        skills.append("Face Verification")

    if row["channel"] == "agent":
        skills.append("Field Onboarding")

    return ", ".join(skills)


tasks["required_skills"] = tasks.apply(
    generate_required_skills,
    axis=1
)
```

```python
tasks[
    [
        "task_type",
        "channel",
        "required_skills"
    ]
].head(15)
```

```python
def determine_priority(row):

    low_id = row["needs_id_review"]
    low_face = row["needs_face_review"]

    if row["kyc_tier"] == "tier3":
        return "High"

    if low_id and low_face:
        return "High"

    if row["kyc_tier"] == "tier2":
        return "Medium"

    if low_id or low_face:
        return "Medium"

    return "Low"


tasks["priority"] = tasks.apply(
    determine_priority,
    axis=1
)
```

## 17. Synthetic Onboarding Agent Profiles

The source onboarding dataset does not contain employee or agent information.

Therefore, synthetic onboarding agent profiles were generated for the prototype. The agents contain attributes required by the task allocation system, including location, skills, experience, workload and availability.

The generated agent records do not represent real employees or organisations.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

NUMBER_OF_AGENTS = 500
```

```python
county_distribution = (
    tasks_ke["county"]
    .value_counts(normalize=True)
    .sort_index()
)

agent_counties = rng.choice(
    county_distribution.index,
    size=NUMBER_OF_AGENTS,
    p=county_distribution.values
)
```

```python
skill_probabilities = {
    "KYC Verification": 0.75,
    "Enhanced KYC Review": 0.35,
    "Identity Verification": 0.55,
    "Face Verification": 0.55,
    "Field Onboarding": 0.25
}
```

```python
def create_agent_skills():

    skills = ["Customer Onboarding"]

    for skill, probability in skill_probabilities.items():

        if rng.random() < probability:
            skills.append(skill)

    return ", ".join(skills)
```

```python
agents = pd.DataFrame({
    "agent_id": [
        f"AGT-{i:04d}"
        for i in range(1, NUMBER_OF_AGENTS + 1)
    ],

    "agent_name": [
        f"Onboarding Agent {i}"
        for i in range(1, NUMBER_OF_AGENTS + 1)
    ],

    "county": agent_counties,

    "experience_years": rng.integers(
        1,
        9,
        NUMBER_OF_AGENTS
    ),

    "skills": [
        create_agent_skills()
        for _ in range(NUMBER_OF_AGENTS)
    ],

    "current_workload": rng.integers(
        0,
        6,
        NUMBER_OF_AGENTS
    )
})
```

```python
agents["availability_status"] = np.where(
    agents["current_workload"] <= 3,
    "Available",
    "Busy"
)
```

## 18. Historical Task Assignment Generation

Historical task-agent assignments were generated because the original onboarding dataset does not contain employee assignment records.

Assignments were generated using task and agent attributes. Agents in the same county as the task were considered first. Skill compatibility was then calculated between the task requirements and the agent's skills.

Where several agents had the same skill compatibility, the agent with fewer previous assignments was selected to encourage workload balance.

These assignments are synthetic and are used to support the collaborative filtering and hybrid recommendation experiments.

```python
from tqdm.auto import tqdm
```

```python
agent_skill_sets = [
    set(skills.split(", "))
    for skills in agents["skills"]
]
```

```python
county_agents = {}

for county in agents["county"].unique():

    county_agents[county] = agents.index[
        agents["county"] == county
    ].to_numpy()
```

```python
assignment_counts = np.zeros(
    NUMBER_OF_AGENTS,
    dtype=int
)

historical_records = []
```

```python
for task in tqdm(
    tasks_ke.itertuples(index=False),
    total=len(tasks_ke)
):

    required = set(
        task.required_skills.split(", ")
    )

    candidates = county_agents[
        task.county
    ]

    skill_scores = np.array([
        len(
            required.intersection(
                agent_skill_sets[i]
            )
        ) / len(required)

        for i in candidates
    ])

    best_skill_score = skill_scores.max()

    best_candidates = candidates[
        np.isclose(
            skill_scores,
            best_skill_score
        )
    ]

    lowest_load = assignment_counts[
        best_candidates
    ].min()

    least_loaded = best_candidates[
        assignment_counts[
            best_candidates
        ] == lowest_load
    ]

    selected_index = int(
        rng.choice(least_loaded)
    )

    assignment_counts[
        selected_index
    ] += 1

    historical_records.append({
        "task_id": task.task_id,
        "task_date": task.task_date,
        "agent_id": agents.loc[
            selected_index,
            "agent_id"
        ],
        "task_profile": task.task_profile,
        "task_complexity": task.task_complexity,
        "skill_match_score": round(
            float(best_skill_score),
            4
        ),
        "location_match": True,
        "completion_time_minutes":
            task.historical_completion_time,
        "task_outcome":
            task.task_outcome
    })
```

```python
assignments = pd.DataFrame(
    historical_records
)

assignments.head()
```

```python
# Sort historical assignments chronologically
assignments = assignments.sort_values(
    ["task_date", "task_id"]
).reset_index(drop=True)

# Total number of records
total_records = len(assignments)

# Split positions
train_end = int(total_records * 0.70)
validation_end = int(total_records * 0.85)

# Create datasets
train_data = assignments.iloc[:train_end].copy()

validation_data = assignments.iloc[
    train_end:validation_end
].copy()

test_data = assignments.iloc[
    validation_end:
].copy()

print("Training:", len(train_data))
print("Validation:", len(validation_data))
print("Testing:", len(test_data))
```

## 20. Agent Performance Calculation

Agent performance was calculated using historical task completion time from the training dataset only.

Using only the training data prevents information from the validation and testing sets from leaking into the model.

Performance was measured relative to the typical completion time for tasks of the same complexity.

```python
# Find the typical completion time for each task complexity
complexity_baselines = (
    train_data
    .groupby("task_complexity")["completion_time_minutes"]
    .median()
)

complexity_baselines
```

```python
train_performance = train_data.copy()

train_performance["baseline_time"] = (
    train_performance["task_complexity"]
    .map(complexity_baselines)
)

train_performance["efficiency_ratio"] = (
    train_performance["baseline_time"]
    / train_performance["completion_time_minutes"]
)

# Convert the ratio to a score between 0 and 1
train_performance["performance_score"] = (
    train_performance["efficiency_ratio"]
    .clip(0.5, 1.5)
    - 0.5
)

train_performance[
    [
        "agent_id",
        "task_complexity",
        "completion_time_minutes",
        "baseline_time",
        "performance_score"
    ]
].head()
```

## 21. Agent Historical Performance

The average performance score and number of historical assignments were calculated for each onboarding agent.

These attributes will later be used by the hybrid recommendation engine when ranking suitable agents.

```python
agent_performance = (
    train_performance
    .groupby("agent_id")["performance_score"]
    .mean()
)

historical_task_count = (
    train_data
    .groupby("agent_id")
    .size()
)
```

```python
agents["performance_score"] = (
    agents["agent_id"]
    .map(agent_performance)
)

agents["historical_tasks"] = (
    agents["agent_id"]
    .map(historical_task_count)
    .fillna(0)
    .astype(int)
)
```

```python
agents["performance_score"] = (
    agents["performance_score"]
    .fillna(agent_performance.mean())
    .round(3)
)
```
