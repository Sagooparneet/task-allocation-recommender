## 6. Create Agent Skills Dataset

The Agent Skills dataset links onboarding agents to the skills they possess.

Each agent may have multiple skills, and each skill may belong to multiple agents.

This relationship is important because the recommender system will later compare agent skills with the skills required by onboarding tasks.

The dataset also includes a proficiency level to represent how strong an agent is in a particular skill.

```python
proficiency_levels = [
    "Beginner",
    "Intermediate",
    "Advanced"
]

agent_skills_data = []

for _, agent in agents_df.iterrows():

    # Every agent receives their specialization as a core skill
    specialization_skill = skills_df[
        skills_df["skill_name"] == agent["specialization"]
    ]

    if not specialization_skill.empty:

        specialization_skill_id = specialization_skill.iloc[0]["skill_id"]

        agent_skills_data.append({
            "agent_id": agent["agent_id"],
            "skill_id": specialization_skill_id,
            "proficiency_level": random.choice(
                ["Intermediate", "Advanced"]
            )
        })

    # Add between 1 and 3 additional skills
    additional_skills = skills_df[
        skills_df["skill_name"] != agent["specialization"]
    ].sample(
        n=random.randint(1, 3),
        random_state=random.randint(1, 10000)
    )

    for _, skill in additional_skills.iterrows():

        agent_skills_data.append({
            "agent_id": agent["agent_id"],
            "skill_id": skill["skill_id"],
            "proficiency_level": random.choice(
                proficiency_levels
            )
        })


agent_skills_df = pd.DataFrame(agent_skills_data)

agent_skills_df.head(10)
```

## 7. Create Synthetic Onboarding Tasks Dataset

The recommender system requires onboarding task information that can be compared with agent profiles.

A synthetic task dataset is created for development while the final industry dataset is still being prepared.

Each task contains:

- Task ID
- Task type
- Description
- Location
- Priority
- Difficulty
- Creation date
- Deadline
- Status

The task types are based on the distributed onboarding activities within the scope of the project, including KYC verification, SIM registration, M-PESA onboarding and account activation.

```python
num_tasks = 2000

task_profiles = {
    "KYC Verification": (
        "Verify customer identity documents and onboarding information."
    ),
    "SIM Registration": (
        "Register and verify a new telecommunications customer."
    ),
    "M-PESA Onboarding": (
        "Assist a customer with M-PESA registration and activation."
    ),
    "Account Activation": (
        "Complete digital account activation for a new customer."
    )
}

priorities = [
    "Low",
    "Medium",
    "High"
]

difficulties = [
    "Low",
    "Medium",
    "High"
]

task_statuses = [
    "Completed",
    "Assigned",
    "Pending"
]

tasks_data = []

start_date = datetime(2026, 1, 1)

for i in range(1, num_tasks + 1):

    task_id = f"T{i:03d}"

    task_type = random.choice(
        list(task_profiles.keys())
    )

    description = task_profiles[task_type]

    location = random.choice(locations)

    priority = random.choices(
        priorities,
        weights=[20, 50, 30]
    )[0]

    difficulty = random.choices(
        difficulties,
        weights=[25, 50, 25]
    )[0]

    created_at = (
        start_date
        + timedelta(
            days=random.randint(0, 250),
            hours=random.randint(0, 23)
        )
    )

    deadline = created_at + timedelta(
        hours=random.choice(
            [24, 48, 72, 96, 120]
        )
    )

    status = random.choices(
        task_statuses,
        weights=[65, 15, 20]
    )[0]

    tasks_data.append({
        "task_id": task_id,
        "task_type": task_type,
        "description": description,
        "location": location,
        "priority": priority,
        "difficulty": difficulty,
        "created_at": created_at,
        "deadline": deadline,
        "status": status
    })


tasks_df = pd.DataFrame(tasks_data)

tasks_df.head()
```

## 8. Create Task Skills Dataset

The Task Skills dataset links each onboarding task to the skills required for successful completion.

Each task is assigned a primary skill based on its task type. Additional supporting skills may also be required depending on the task.

The required proficiency level is influenced by task difficulty.

This dataset will later support:

- Content-based agent-to-task matching
- Knowledge graph relationships
- Recommendation scoring
- Explanation generation

```python
# Map each task type to its primary required skill
primary_skill_map = {
    "KYC Verification": "S001",
    "SIM Registration": "S002",
    "M-PESA Onboarding": "S003",
    "Account Activation": "S004"
}

# Supporting skills that may also be required
supporting_skill_map = {
    "KYC Verification": [
        "S005",  # Customer Service
        "S006",  # Identity Document Verification
        "S007"   # Digital Registration
    ],

    "SIM Registration": [
        "S005",  # Customer Service
        "S006",  # Identity Document Verification
        "S007"   # Digital Registration
    ],

    "M-PESA Onboarding": [
        "S005",  # Customer Service
        "S007",  # Digital Registration
        "S008"   # Mobile Financial Services
    ],

    "Account Activation": [
        "S005",  # Customer Service
        "S006",  # Identity Document Verification
        "S007"   # Digital Registration
    ]
}

# Map task difficulty to the proficiency expected
difficulty_level_map = {
    "Low": "Beginner",
    "Medium": "Intermediate",
    "High": "Advanced"
}

task_skills_data = []

for _, task in tasks_df.iterrows():

    task_id = task["task_id"]
    task_type = task["task_type"]
    difficulty = task["difficulty"]

    # Primary skill
    primary_skill_id = primary_skill_map[task_type]

    task_skills_data.append({
        "task_id": task_id,
        "skill_id": primary_skill_id,
        "required_level": difficulty_level_map[difficulty]
    })

    # Add 0 to 2 supporting skills
    number_of_supporting_skills = random.randint(0, 2)

    if number_of_supporting_skills > 0:

        selected_supporting_skills = random.sample(
            supporting_skill_map[task_type],
            k=number_of_supporting_skills
        )

        for skill_id in selected_supporting_skills:

            task_skills_data.append({
                "task_id": task_id,
                "skill_id": skill_id,
                "required_level": random.choice(
                    ["Beginner", "Intermediate"]
                )
            })


task_skills_df = pd.DataFrame(task_skills_data)

task_skills_df.head(10)
```

## 9. Create Historical Assignments Dataset

The Historical Assignments dataset represents previous task allocation interactions between onboarding agents and onboarding tasks.

This information is important for the collaborative filtering component because it provides evidence of which agents previously handled different types of onboarding tasks and how well they performed.

Assignments are generated only for tasks whose status is either `Assigned` or `Completed`.

Where possible, tasks are allocated to agents who possess the primary skill required by the task. Location, workload, experience and performance are also considered when generating realistic development records.

The dataset contains:

- Assignment ID
- Task ID
- Agent ID
- Assignment status
- Assignment date
- Completion date
- Completion time
- Performance rating

```python
# Map proficiency levels to numerical values
proficiency_scores = {
    "Beginner": 1,
    "Intermediate": 2,
    "Advanced": 3
}

historical_assignments_data = []

# Only Assigned and Completed tasks should have assignment records
assignable_tasks = tasks_df[
    tasks_df["status"].isin(
        ["Assigned", "Completed"]
    )
].copy()


for assignment_number, (_, task) in enumerate(
    assignable_tasks.iterrows(),
    start=1
):

    task_id = task["task_id"]
    task_type = task["task_type"]
    task_location = task["location"]
    task_difficulty = task["difficulty"]

    # Get the primary skill required by the task
    primary_skill_id = primary_skill_map[
        task_type
    ]

    # Find agents who possess the primary skill
    skilled_agent_ids = agent_skills_df[
        agent_skills_df["skill_id"]
        == primary_skill_id
    ]["agent_id"].tolist()

    candidate_agents = agents_df[
        agents_df["agent_id"].isin(
            skilled_agent_ids
        )
    ].copy()

    # Prefer agents in the same location where possible
    same_location_agents = candidate_agents[
        candidate_agents["location"]
        == task_location
    ]

    if (
        not same_location_agents.empty
        and random.random() < 0.70
    ):
        candidate_agents = same_location_agents

    # Give stronger agents a slightly higher chance
    # of being selected in the historical data
    selection_weights = (
        candidate_agents["performance_score"]
        + candidate_agents["experience_years"] * 2
        - candidate_agents["workload"] * 2
    )

    selected_agent = candidate_agents.sample(
        n=1,
        weights=selection_weights,
        random_state=random.randint(
            1,
            100000
        )
    ).iloc[0]

    agent_id = selected_agent["agent_id"]

    # Get proficiency in the primary skill
    skill_record = agent_skills_df[
        (
            agent_skills_df["agent_id"]
            == agent_id
        )
        &
        (
            agent_skills_df["skill_id"]
            == primary_skill_id
        )
    ].iloc[0]

    proficiency_level = skill_record[
        "proficiency_level"
    ]

    assigned_at = (
        task["created_at"]
        + timedelta(
            hours=random.randint(1, 12)
        )
    )

    assignment_status = task["status"]

    # Completed assignments contain outcome information
    if assignment_status == "Completed":

        difficulty_time_ranges = {
            "Low": (2, 6),
            "Medium": (4, 12),
            "High": (8, 24)
        }

        minimum_hours, maximum_hours = (
            difficulty_time_ranges[
                task_difficulty
            ]
        )

        completion_time_hours = round(
            random.uniform(
                minimum_hours,
                maximum_hours
            ),
            2
        )

        completed_at = (
            assigned_at
            + timedelta(
                hours=completion_time_hours
            )
        )

        # Convert the agent's performance score
        # into an approximate 1-5 rating
        rating = (
            selected_agent["performance_score"]
            / 20
        )

        # Adjust based on skill proficiency
        if proficiency_level == "Advanced":
            rating += 0.20

        elif proficiency_level == "Beginner":
            rating -= 0.20

        # Small benefit for location match
        if (
            selected_agent["location"]
            == task_location
        ):
            rating += 0.10

        # Add small random variation
        rating += np.random.normal(
            0,
            0.15
        )

        # Keep rating between 1 and 5
        performance_rating = round(
            min(
                5.0,
                max(
                    1.0,
                    rating
                )
            ),
            2
        )

    else:

        completed_at = pd.NaT
        completion_time_hours = np.nan
        performance_rating = np.nan

    historical_assignments_data.append({
        "assignment_id":
            f"AS{assignment_number:04d}",

        "task_id":
            task_id,

        "agent_id":
            agent_id,

        "status":
            assignment_status,

        "assigned_at":
            assigned_at,

        "completed_at":
            completed_at,

        "completion_time_hours":
            completion_time_hours,

        "performance_rating":
            performance_rating
    })


historical_assignments_df = pd.DataFrame(
    historical_assignments_data
)

historical_assignments_df.head(10)
```
